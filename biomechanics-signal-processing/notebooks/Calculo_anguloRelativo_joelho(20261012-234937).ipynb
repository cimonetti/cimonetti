```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1 Crie um arquivo de função que faça o restante dos comandos

data=np.loadtxt('dadosciclismo.csv',delimiter=',')
```

```python
# Definindo os segmentos
def calculaAngulo_joelho (data):
  segmentoCoxa = data[:,1:4] - data[:,4:7]
  segmentoPerna = data[:,7:10] - data[:,4:7]
  # 2- Calcule  ângulo relativo do joelho em cada estante de tempo

  tempo = data[:,0]
  angulo = []

  for i in range(len(segmentoCoxa)):
    denominador = np.linalg.norm(segmentoCoxa[i,:])*np.linalg.norm(segmentoPerna[i,:])
    if denominador == 0:
      angulo_frame = np.nan
    else:
      angulo_frame = np.arccos(np.clip((np.dot(segmentoCoxa[i,:],segmentoPerna[i,:]))/
      denominador, -1, 1))*180/np.pi
    angulo.append(angulo_frame)
  return tempo, angulo
```

```python
# Apresente um gráfico dos valores de ângulo em função de tempo
tempo, angulo = calculaAngulo_joelho(data)

plt.plot(tempo,angulo, '-b')
plt.title('Ângulo do joelho x Tempo')
plt.xlabel('Tempo (s)')
plt.ylabel('Ângulo(graus)')
plt.show()
```

```python
def detect_picos_vales(angulo):
    picos = []
    vales = []

    for i in range(1, len(angulo)-1):
        anterior = angulo[i-1]
        atual = angulo[i]
        proximo = angulo[i+1]

        if atual > anterior and atual > proximo:
            picos.append(i)
        elif atual < anterior and atual < proximo:
            vales.append(i)


    vales_filtrados = [vales[0]] if vales else []

    for vale in vales[1:]:
        if vale - vales_filtrados[-1] >= 150:
            vales_filtrados.append(vale)

    return picos, vales_filtrados




def calcula_ciclos(vales_filtrados):
    ciclos = []

    for i in range(len(vales_filtrados)-1):
        inicio = vales_filtrados[i]
        fim = vales_filtrados[i+1]
        ciclos.append((inicio, fim))

    return ciclos







def min_max_ciclos(angulo, ciclos):
    minimos = []
    maximos = []

    for inicio, fim in ciclos:
        trecho = angulo[inicio:fim + 1]
        minimo = min(trecho)
        maximo = max(trecho)

        minimos.append(minimo)
        maximos.append(maximo)

    return minimos, maximos


```

```python
picos, vales = detect_picos_vales(angulo)
print("Picos:", picos)
print("Vales filtrados:", vales)

ciclos = calcula_ciclos(vales)
print("Ciclos:", ciclos)

minimos, maximos = min_max_ciclos(angulo, ciclos)
print("Mínimos por ciclo:", minimos)
print("Máximos por ciclo:", maximos)
resultados = np.column_stack((minimos, maximos))
np.savetxt("resultados.txt", resultados, fmt="%.2f", header="Minimo Maximo")
```
