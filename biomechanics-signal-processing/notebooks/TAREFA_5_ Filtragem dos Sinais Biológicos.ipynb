```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal


frs = np.loadtxt('frs.txt')

tempo = frs[:,0]
fx = frs[:,1]
fy = frs[:,2]
fz = frs[:,3]


plt.figure(figsize=(10,6))

plt.subplot(3,1,1)
plt.plot(tempo, fx)
plt.title('Fx bruto')

plt.subplot(3,1,2)
plt.plot(tempo, fy)
plt.title('Fy bruto')

plt.subplot(3,1,3)
plt.plot(tempo, fz)
plt.title('Fz bruto')

plt.tight_layout()
plt.show()


fs = 1000

for sinal, nome in zip([fx, fy, fz], ['Fx','Fy','Fz']):

    n = len(sinal)
    freq = np.fft.rfftfreq(n, d=1/fs)
    fft_vals = np.abs(np.fft.rfft(sinal))**2

    forca_total = np.sum(fft_vals)

    soma = np.cumsum(fft_vals)

    freq_mediana = freq[np.where(soma >= forca_total/2)[0][0]]
    freq_pred = freq[np.argmax(fft_vals)]
    freq_95 = freq[np.where(soma >= 0.95*forca_total)[0][0]]

    print(nome)
    print("Mediana:", freq_mediana)
    print("Predominante:", freq_pred)
    print("F95:", freq_95)
    print()


Freq_aquis = 1000

# FX
residuo_fx = []
fc_fx = []

for i in np.arange(0.5,20,0.5):
    b,a = signal.butter(4, i/(Freq_aquis/2))
    filtrado = signal.filtfilt(b,a,fx)
    res = np.sqrt(np.mean((filtrado - fx)**2))
    residuo_fx.append(res)
    fc_fx.append(i)

plt.figure()
plt.plot(fc_fx, residuo_fx)
plt.title('Residual Fx')
plt.show()


# FY
residuo_fy = []
fc_fy = []

for i in np.arange(0.5,20,0.5):
    b,a = signal.butter(4, i/(Freq_aquis/2))
    filtrado = signal.filtfilt(b,a,fy)
    res = np.sqrt(np.mean((filtrado - fy)**2))
    residuo_fy.append(res)
    fc_fy.append(i)

plt.figure()
plt.plot(fc_fy, residuo_fy)
plt.title('Residual Fy')
plt.show()


# FZ
residuo_fz = []
fc_fz = []

for i in np.arange(0.5,20,0.5):
    b,a = signal.butter(4, i/(Freq_aquis/2))
    filtrado = signal.filtfilt(b,a,fz)
    res = np.sqrt(np.mean((filtrado - fz)**2))
    residuo_fz.append(res)
    fc_fz.append(i)

plt.figure()
plt.plot(fc_fz, residuo_fz)
plt.title('Residual Fz')
plt.show()


fc = 4

b,a = signal.butter(4, fc/(Freq_aquis/2))

fx_f = signal.filtfilt(b,a,fx)
fy_f = signal.filtfilt(b,a,fy)
fz_f = signal.filtfilt(b,a,fz)


plt.figure(figsize=(10,6))

plt.subplot(3,1,1)
plt.plot(tempo, fx, 'r')
plt.plot(tempo, fx_f)
plt.title('Fx')

plt.subplot(3,1,2)
plt.plot(tempo, fy, 'r')
plt.plot(tempo, fy_f)
plt.title('Fy')

plt.subplot(3,1,3)
plt.plot(tempo, fz, 'r')
plt.plot(tempo, fz_f)
plt.title('Fz')

plt.tight_layout()
plt.show()

plt.figure()
plt.plot(fc_fx, residuo_fx)
plt.title('Residual Fx')

plt.figure()
plt.plot(fc_fy, residuo_fy)
plt.title('Residual Fy')

plt.figure()
plt.plot(fc_fz, residuo_fz)
plt.title('Residual Fz')
```

```python
plt.figure(figsize=(12, 8))

for i, (sinal, nome) in enumerate(zip([fx, fy, fz], ['Fx', 'Fy', 'Fz'])):
    n = len(sinal)
    freq = np.fft.rfftfreq(n, d=1/fs)
    fft_vals = np.abs(np.fft.rfft(sinal))

    plt.subplot(3, 1, i+1)
    plt.plot(freq, fft_vals)
    plt.title(f'Análise Espectral de {nome}')
    plt.xlabel('Frequência (Hz)')
    plt.ylabel('Magnitude da FFT')
    plt.xlim(0, 20)
    plt.grid(True)

plt.tight_layout()
plt.show()
```
